```python
#não temos uma probabilidade, então usaremos poisson prob exata e uma média com base no observado pelo total

#@title **junção agregada das duas bases**

# bibliotecas
import pandas as pd
import numpy as np
from statsmodels.stats.proportion import proportions_ztest
from scipy import stats

# export no drive
from google.colab import drive
drive.mount('/content/drive')

# caminhos
CAMINHO_POLTRA = "/content/drive/MyDrive/collab/POLTRA_25.csv"
CAMINHO_INDICA = "/content/drive/MyDrive/collab/indica.csv"
STATE_TARGET = 'Arizona' # estado do arizona, que sera utilizado na analise

# carregando as duas bases de cima
try:
    poltra = pd.read_csv(CAMINHO_POLTRA)
    indica = pd.read_csv(CAMINHO_INDICA)
    print("bases carregadas com sucesso.")
except FileNotFoundError as e:
    # Se der ruim, mostra o erro e para
    print(f"ERRO CRÍTICO: Um dos arquivos não foi encontrado. Erro: {e}")


# formatando a data e filtrando as colunas que buscamos
poltra['arrest_dt'] = pd.to_datetime(poltra['arrest_dt'], errors='coerce')
indica['Year'] = pd.to_datetime(indica['Time Period Start Date']).dt.year

COL_GENERO = 'arrestee_genderTranslation'
VALOR_MULHER = 'Female'


# pegando o indicador de saude mental do
# filtrando so as linhas q se referem ao arizona e que possuam casos de ansiedade ou depressão
indica_filtrada = indica[
    (indica['State'] == STATE_TARGET) &
    (indica['Indicator'].str.contains('Anxiety|Depression', case=False, na=False))
]


#verifcação de se a pessoa detida é mulher
poltra['is_female'] = (poltra[COL_GENERO] == VALOR_MULHER)

# vendo o total de detençao e quantas foram mulheres
poltra_agg = poltra.groupby(poltra['arrest_dt'].dt.year).agg(
    total_detencoes=('primary_key', 'count'), # Conta o total de detenções
    mulheres_detidas=('is_female', 'sum')     # Soma o número de mulheres detidas
).reset_index().rename(columns={'arrest_dt': 'Year'})

# calculo de proporção
poltra_agg['prop_mulheres'] = poltra_agg['mulheres_detidas'] / poltra_agg['total_detencoes']
poltra_agg['State'] = STATE_TARGET


# pega os valores do indicador por ano e estado e junta com a proporção de detenção
dados_combinados = pd.merge(
    poltra_agg,
    indica_filtrada,
    on=['Year', 'State'],
    how='inner'
)

# criação da base final mostrando a proporção das mulheres e o indicador media
dados_finais = dados_combinados.groupby('Year').agg(
    prop_mulheres=('prop_mulheres', 'mean'),
    indicador_media=('Value', 'mean')
).reset_index()

print("\n--- RESUMO DA BASE FINAL POR ANO (dados_finais) ---")
print("Dados prontos para análise:")
print(dados_finais)
```

```python
#@title **os percentuais de um ano foram maior que o outro? - 2022/23**

#filtrar os percentuais desejados de cada ano no arizona
#acho que usar o teste do shapiro para comparar e checar o p-value

#@title **os percentuais de um ano foram maior que o outro? (De muylheres detidas q tenham o indicador) - 2022/23**

# teste estatístico para ver se a média do indicador (Ansiedade/Depressão) subiu de 2022 para 2023.

indicador_2023 = dados_combinados[dados_combinados['Year'] == 2023]['Value']
indicador_2022 = dados_combinados[dados_combinados['Year'] == 2022]['Value']

alpha = 0.05
print("-" * 50)
print("os percentuais de um ano foram maior que o outro? (De muylheres detidas q tenham o indicador) - 2022/23")

# verifca o numero de amostra (nesse caso, pra validar tem q ter mais de 2)
if len(indicador_2023) > 1 and len(indicador_2022) > 1:

    # fazemos o teste t ed amostras independentes
    t_stat, p_value_t = stats.ttest_ind(
        indicador_2023.dropna(),
        indicador_2022.dropna(),
        alternative='greater',
        equal_var=False
    )

    media_2023 = indicador_2023.mean()
    media_2022 = indicador_2022.mean()

    print(f"-> Média do Indicador em 2023: {media_2023:.3f} (N={len(indicador_2023)} amostras)")
    print(f"-> Média do Indicador em 2022: {media_2022:.3f} (N={len(indicador_2022)} amostras)")
    print(f"-> Estatística T (medida da diferença): {t_stat:.3f}")
    print(f"-> P-valor (prob. de erro): {p_value_t:.5f}")

    # conclusão final da análise
    if p_value_t < alpha:
        print(f" conclusão final: O INDICADOR EM 2023 É ESTATISTICAMENTE MAIOR que em 2022 (p-valor < {alpha}).")
    else:
        print(f"conclusão final: Não há evidência estatística de que o INDICADOR TENHA AUMENTADO de 2022 para 2023.")
else:
    print("os dados nn foram suficientes pra fazer a comparação, necessitando usar outro método, comparando as medias por exemplo.")

print("-" * 50)
```
